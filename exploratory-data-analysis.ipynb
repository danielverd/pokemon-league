```python
import polars as pl
df = pl.read_csv("PokemonLeague_statsheet_R9.tsv", separator="\t")
print(df.columns)
```

```python
# SELECT Pokemon, sum(Kills) FROM df
# GROUP BY Pokemon DESC

result = pl.sql("""
    SELECT Pokemon, SUM(Kills) as TotalKills
    FROM df 
    GROUP BY Pokemon
    ORDER BY TotalKills DESC
""").collect()

pl.Config.set_tbl_rows(-1)
pl.Config.set_tbl_cols(-1)
print(result)
```

```python
# SELECT Pokemon, sum(Kills) FROM df
# GROUP BY Pokemon DESC

result = pl.sql("""
    SELECT PokemonName, SUM(TurnsIn) as TotalTurns
    FROM df 
    GROUP BY PokemonName
    ORDER BY TotalTurns DESC
""").collect()
```

```python
print(result)
```

```python
result = pl.sql("""
    SELECT PlayerName, SUM(Kills) as TotalKills, COUNT(PlayerName)/6 as GamesPlayed, 1.0 * SUM(Kills)/(COUNT(PlayerName)/6) as KPG 
    FROM df 
    GROUP BY PlayerName
    ORDER BY KPG DESC
""").collect()
```

```python
print(result)
```

```python
result = pl.sql("""
    SELECT PokemonName, SUM(Kills) as TotalKills, 1.0 * SUM(Kills)/(8) as KPM
    FROM df 
    GROUP BY PokemonName
    ORDER BY KPM DESC
""").collect()
print(result)
```
